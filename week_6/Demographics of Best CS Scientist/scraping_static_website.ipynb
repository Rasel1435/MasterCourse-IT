```python
import requests
import pandas as pd
import bs4

url = "https://en.wikipedia.org/wiki/Red_states_and_blue_states"
web_page = bs4.BeautifulSoup(requests.get(url, {}).text, "html.parser")
table = web_page.find_all(name="table", attrs={"class": "wikitable"})
table_rows = table[-1].find_all(name='tr')
table_columns = table_rows[0].find_all(name='th')
table_row_name = [row_name.text.replace("\n", "") for row_name in table_rows[0].find_all(name='th')]
table_content = [row_name.text.replace("\n", "").strip() for row_name in table_rows[2].find_all(name='td')]
table_content
```

```python
x_dict = {}
for idx, row_name in enumerate(table_rows[2].find_all(name='td')):
    key = table_columns[idx].text.strip()
    value = row_name.text.replace("\n", "").strip()
    x_dict[key] = value
x_dict
```

```python
x_dict = {}
for idx, row_name in enumerate(table_rows[2].find_all(name='td')):
    x_dict[table_columns[idx].text.strip()] = row_name.text.replace("\n", "").strip()
x_dict
```

```python
contents = []
for row_id in range(2, len(table_rows)):
    table_content = {table_columns[idx].text.strip() :  row_name.text.replace("\n", "").strip() for idx, row_name in enumerate(table_rows[row_id].find_all(name='td'))}
    contents.append(table_content)
```

```python
contents = []
for row_id in range(2, len(table_rows)):
    contents.append({table_columns[idx].text.strip() :  row_name.text.replace("\n", "").strip() for idx, row_name in enumerate(table_rows[row_id].find_all(name='td'))})
```

```python
contents = [{table_columns[idx].text.strip() :  row_name.text.replace("\n", "").strip() for idx, row_name in enumerate(table_rows[row_id].find_all(name='td'))} for row_id in range(2, len(table_rows))]
```

```python
contents[-1]
```

```python
# Extract clean column names
table_columns = [col.text.strip() for col in table_columns]

# Extract table data row-wise
contents = [
    [row_name.text.replace("\n", "").strip() for row_name in table_rows[row_id].find_all(name='td')]
    for row_id in range(2, len(table_rows))
]

# Create DataFrame correctly
df = pd.DataFrame(data=contents, columns=table_columns)
df.head()
```
