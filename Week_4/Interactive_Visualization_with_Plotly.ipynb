```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cufflinks as cf
import plotly.express as px
import plotly.graph_objects as go
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
from plotly.offline import init_notebook_mode
init_notebook_mode(connected=True)

cf.go_offline()
```

### Load The Data

```python
df = pd.read_csv('./data/computer_science_scientists.csv')
df.head()
```

### Line Chart
Signle Line Chart

```python
px.line(df, x='World Rank', y='#DBLP').show(render='colab')
```

#### Multiple Line Chart 

```python
px.line(df, x='World Rank', y=['#DBLP', 'National Rank', 'H-Index'], title='W-RANK VS ALL').show(render='colab')
```

### Styling Line Chart

```python
fig = go.Figure()

fig.add_trace(go.Scatter(x=df['World Rank'], y=df['H-Index'],
                         mode = 'lines', name='H-Index'))
fig.add_trace(go.Scatter(x=df['World Rank'], y=df['National Rank'],
                         mode = 'lines+markers', name='National Rank'))
fig.add_trace(go.Scatter(x=df['World Rank'], y=df['#DBLP'],
                         mode = 'lines+markers', name='#DBLP',
                         line=dict(color='blue', width=2, dash='dashdot')))


fig.update_layout(
    title = 'WorldRank VS ALL',
    xaxis_title = 'World Rank',
    yaxis_title = 'Value'
)

fig.show(render='colab')
```

### Bar Plots

```python
country_cnt = df['Country'].value_counts().to_dict()
countries = list(country_cnt.keys())
cnt = list(country_cnt.values())

px.bar(x=countries, y=cnt, labels={'x': 'Country', 'y':'Count'}).show(render='colab')
```

### Histogram

```python
px.histogram(df, x='#DBLP', nbins=100).show(render='colab')
```

### Box Plot
Single Box Plot

```python
px.box(df, x='#DBLP', points='all').show(render='colab')
```

Multiple BoxPlots

```python
px.box(df, y=['#DBLP', 'H-Index'], points='all').show(render='colab')
```

Violin Plot

```python
px.violin(df, y=['#DBLP', 'H-Index'], points='all').show(render='colab')
```

### Scatter Plot

```python
px.scatter(df,
    x='Citations',
    y='#DBLP',
    color='Country',
    size='H-Index',
    hover_data=['World Rank', 'Name', 'Affiliation']).show(render='colab')
```
