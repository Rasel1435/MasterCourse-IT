```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
sns.set()
sns.set(style='darkgrid')
```

Load The Data

```python
df = pd.read_csv('/backup/ML/MasterCourse-IT/week_4/data/computer_science_scientists.csv')
df.head()
```

# Basic Plot

Line Chart

```python
plt.rcParams['figure.figsize'] = plt.rcParamsDefault['figure.figsize']
plt.rcParamsDefault['figure.figsize'] #Default size
```

```python
plt.rcParams['figure.figsize'] = [15,5]
sns.lineplot(x= "World Rank", y= "#DBLP", data=df)
```

# Bar Plot

```python
plt.rcParams['figure.figsize'] = (20,10)
country_cnt = df['Country'].value_counts().to_dict()
countries = list(country_cnt.keys())
cnt = list(country_cnt.values())
ax = sns.barplot(x=countries, y=cnt)
ax.set_title("Country Presence")
ax.set_xticklabels(ax.get_xticklabels(), rotation=90)
```

# Histogram

```python
sns.histplot(df['#DBLP'], kde=True)
```

# Box Plot

```python
sns.boxplot(data=df['#DBLP'])
```

# Violin Plot

```python
sns.violinplot(data=df[['#DBLP', 'H-Index']])
```

# Scatter Plots

```python
sns.relplot(x='#DBLP', y='World Rank', data=df[:200], kind='scatter', height=5, aspect=2);
```

```python
sns.relplot(x='Citations', y='World Rank', data=df[:200], kind='scatter', height=5, aspect=2);
```

```python
sns.relplot(x='Citations', y='#DBLP', data=df[:200], hue='Country', height=5, aspect=2);
```

# Bubble Plot

```python
sns.relplot(x='Citations', y='#DBLP', size='World Rank', data=df[:200], hue='Country', height=10, aspect=2);
```

# SubPlot

```python
sns.relplot(x='Citations', y='#DBLP', col='Country', hue='World Rank', size='World Rank',
            data=df[:200],col_wrap=2, height=5, aspect=2);
```

# Advance 
<ul>Categorical scatter plot</ul>
<ul>Strip Plot</ul>

```python
sns.catplot(x='#DBLP', y='Country', kind='strip', data=df[:200], height=7, aspect=2);
```

Swarm Plot

```python
sns.catplot(x='#DBLP', y='Country', kind='swarm', data=df[:200], height=7, aspect=2);
```

### Categorical Distribution Plots
Box Plot

```python
sns.catplot(x='Citations', y='Country', kind='box', data=df, height=10, aspect=2);
```

Violin Plot

```python
sns.catplot(x='H-Index', y='Country', kind='violin', data=df, height=10, aspect=2);
```

### Categorical Estimate Plots
Bar Plot

```python
sns.catplot(x='#DBLP', y='Country', kind='bar', data=df, height=10, aspect=2);
```

### Density Plot
KDE Plot - Kernel density estimate plots

```python
plt.rcParams['figure.figsize'] = [15,5]
sns.kdeplot(data=df['#DBLP'], shade=True);
```

```python
plt.rcParams['figure.figsize'] = [15,5]
sns.kdeplot(data=df['Citations'], shade=True);
```

```python
plt.rcParams['figure.figsize'] = [15,5]
sns.kdeplot(x=df['#DBLP'], y=df['Citations'], shade=True);
```

### Pair Plots
Can Visualize multidimensional Relationships

```python
data = df[["#DBLP", 'Citations', 'H-Index', 'Country', 'World Rank']]
data.head()
```

```python
sns.pairplot(data, hue='Country', height=5)
```
