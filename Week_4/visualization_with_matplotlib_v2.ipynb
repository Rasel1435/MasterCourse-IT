# Matplotlib

Import Necessary Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

Load The Data

```python
df = pd.read_csv('./data/computer_science_scientists.csv')
df.head()
```

# Line Chart

```python
country_cnt = df['Country'].value_counts().to_dict()
countries = list(country_cnt.keys())
cnt = list(country_cnt.values())
plt.rcParams["figure.figsize"] = (15, 5)
plt.title('Presence of Countries')
plt.plot(cnt, 'g--')
plt.xticks(ticks = range(len(countries)), labels=countries, rotation='vertical')
plt.show()
```

# Bar Plot

```python
country_cnt = df['Country'].value_counts().to_dict()
countries = list(country_cnt.keys())
cnt = list(country_cnt.values())

plt.rcParams['figure.figsize'] = [28,18]
plt.title('Presence of Countries')
plt.bar(countries, cnt, color=['red', 'blue', 'green', 'lightblue'])
plt.xticks(ticks=range(len(countries)), labels=countries, rotation='vertical')
plt.show()
```

# Histogram

```python
plt.rcParams['figure.figsize'] = [10,5]
plt.title('#DBLP Distribution')
plt.xlabel('#DBLP')
plt.ylabel('Frequency')
plt.hist(df['#DBLP'], bins=50, color='lightblue')
plt.show()
```

# Box Plot

```python
outliers = dict(markerfacecolor='r', marker='D')
plt.title('#DBLP Distribution')
plt.boxplot(df['#DBLP'], labels=['#DBLP'], flierprops=outliers)
plt.show()
```

# Multiple Box Plot in One

```python
data = df[['#DBLP', 'H-Index']]
outliers = dict(markerfacecolor='g', marker = 'D')
plt.rcParams['figure.figsize'] = [20,10]
fig, ax = plt.subplots()
plt.boxplot(data.values, labels=['#DBLP', 'H-Index'], flierprops=outliers)
plt.show()
```

```python
data = df[['#DBLP', 'H-Index', 'Citations']]
outliers = dict(markerfacecolor='b', marker = 'o')
plt.rcParams['figure.figsize'] = [20,10]
fig, ax = plt.subplots()
plt.boxplot(data.values, labels=['#DBLP', 'H-Index', 'Citations'], flierprops=outliers)
plt.show()
```

# Violin Plots

```python
data = df[['#DBLP', 'H-Index']]
plt.rcParams['figure.figsize'] = [20,10]
fig, ax = plt.subplots()
plt.xticks(ticks=[1,2], labels=['#DBLP', 'H-Index'])
plt.violinplot(data.values)
plt.show()
```

# Scatter Plots
shows relative distributions of two variables

```python
plt.xlabel('#DBLP')
plt.ylabel('World Rank')

plt.scatter(df['#DBLP'], df['World Rank'])
```

```python
plt.xlabel('citations')
plt.ylabel('World Rank')

plt.scatter(df['Citations'], df['World Rank'])
```

# Bubble Plots
Relative distribution among 3 variables

```python
plt.xlabel('Citations')
plt.ylabel('#DBLP')

plt.scatter(df['Citations'][:200], df['#DBLP'][:200], s=df['World Rank'][:200]) # s is the size of bubble
```
